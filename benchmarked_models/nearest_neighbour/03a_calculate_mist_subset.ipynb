Import libraries

```python
import os 
import pickle
from pathlib import Path
```

Helper Functions

```python
def load_pickle(path):
    
    with open(path, 'rb') as f:
        return pickle.load(f)
```

Compute the results on the subset of MIST

```python
datasets = ["NPLIB1", "massspecgym"]
results_dir = Path("/data/rbg/users/klingmin/projects/ML_MS_analysis/results/FP_prediction/mist")

for d in datasets: 

    current_dir = results_dir / d 

    for ckpt in os.listdir(current_dir):
        
        split = ckpt.split("_")[-1]
        if split == "LS": continue
        nn_path = Path(f"/data/rbg/users/klingmin/projects/ML_MS_analysis/results/nearest_neighbour/nn_sim/{d}_{split}.pkl")
        nn = load_pickle(nn_path)[0]

        results_path = current_dir / ckpt / "test_results.pkl"
        results = load_pickle(results_path)
        total_jaccard, count = 0, 0 
        
        for k, v in results.items():

            if str(k) in nn: 
                total_jaccard += v["jaccard"]
                count += 1 

        print(ckpt, len(results), round(total_jaccard / count, 3), count)
        print()
```
